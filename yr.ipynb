```python
import pandas as pd

# === INTERNET USAGE ===
internet = pd.read_csv("individualsusingtheinternet.csv")
internet = internet[(internet["Year"] >= 2000) & (internet["Year"] <= 2023)]
internet = internet[["Entity", "Year", "Individuals using the Internet (% of population)"]]
internet.rename(columns={"Individuals using the Internet (% of population)": "Internet_Usage"}, inplace=True)

# === HDI ===
hdi = pd.read_csv("human-development-index.csv")
hdi = hdi[(hdi["Year"] >= 2000) & (hdi["Year"] <= 2023)]
hdi = hdi[["Entity", "Year", "Human Development Index"]]
hdi.rename(columns={"Human Development Index": "HDI"}, inplace=True)

# === GDP ===
gdp = pd.read_excel("GDP.xls", skiprows=3)
gdp_long = gdp.melt(id_vars=["Country Name"], var_name="Year", value_name="GDP")
gdp_long["Year"] = pd.to_numeric(gdp_long["Year"], errors="coerce")
gdp_long = gdp_long[(gdp_long["Year"] >= 2000) & (gdp_long["Year"] <= 2023)]
gdp_long.rename(columns={"Country Name": "Entity"}, inplace=True)

# === MOBILE ===
mobile = pd.read_excel("Mobilesubscriptionsper100.xls", skiprows=3)
mobile_long = mobile.melt(id_vars=["Country Name"], var_name="Year", value_name="Mobile_Subscriptions")
mobile_long["Year"] = pd.to_numeric(mobile_long["Year"], errors="coerce")
mobile_long = mobile_long[(mobile_long["Year"] >= 2000) & (mobile_long["Year"] <= 2023)]
mobile_long.rename(columns={"Country Name": "Entity"}, inplace=True)

# === MERGE EVERYTHING ===
merged_long = (
    internet.merge(hdi, on=["Entity", "Year"], how="outer")
            .merge(gdp_long, on=["Entity", "Year"], how="outer")
            .merge(mobile_long, on=["Entity", "Year"], how="outer")
)

# === SORT AND SAVE ===
merged_long = merged_long.sort_values(by=["Entity", "Year"]).reset_index(drop=True)
merged_long.to_excel("Global_Data_With_Year.xlsx", index=False)

print("✅ Done! 'Global_Data_With_Year.xlsx' created successfully.")
print(merged_long.head())
```
