```python
#Cleaning the GDP dataset and calculating average GDP per country from 2000 to 2024, then saving to Excel.
import pandas as pd

# 1️⃣ Load the GDP dataset
gdp = pd.read_excel("GDP.xls", skiprows=3)  # skip first 3 rows (typical for World Bank data)

# 2️⃣ Show first few rows and column names to confirm
print(gdp.head())
print(gdp.columns)

# 3️⃣ Keep only relevant columns
# Assuming your columns look like: ['Country Name', 'Country Code', '2000', '2001', ..., '2024']
# We’ll select the country and year columns
cols_to_keep = ['Country Name'] + [str(year) for year in range(2000, 2025)]
gdp = gdp[cols_to_keep]

# 4️⃣ Convert the year columns to numeric (in case some are text)
gdp = gdp.apply(pd.to_numeric, errors='ignore')

# 5️⃣ Compute the average GDP across all years for each country
gdp['Avg_GDP'] = gdp.loc[:, '2000':'2023'].mean(axis=1)

# 6️⃣ Keep only country name and average GDP
gdp_avg = gdp[['Country Name', 'Avg_GDP']]

# 7️⃣ Rename for consistency
gdp_avg.rename(columns={'Country Name': 'Entity'}, inplace=True)

# 8️⃣ Save to Excel
gdp_avg.to_excel("Avg_GDP.xlsx", index=False)

print("✅ Done! 'Avg_GDP.xlsx' created with average GDP per country (2000–2024).")
```

```python
#Cleaning the mobile subscriptions dataset and calculating average mobile subscriptions per country from 2000 to 2023, then saving to Excel.
import pandas as pd

# 1️⃣ Load the dataset
mobile = pd.read_excel("Mobilesubscriptionsper100.xls", skiprows=3)

# 2️⃣ Preview the first few rows and column names
print(mobile.head())
print(mobile.columns)

# 3️⃣ Keep only the relevant columns (Country + 2000–2023)
cols_to_keep = ['Country Name'] + [str(year) for year in range(2000, 2024)]
mobile = mobile[cols_to_keep]

# 4️⃣ Convert year columns to numeric (ignore non-numeric values)
mobile = mobile.apply(pd.to_numeric, errors='ignore')

# 5️⃣ Compute average mobile subscriptions (2000–2023)
mobile['Avg_Mobile_Subscriptions'] = mobile.loc[:, '2000':'2023'].mean(axis=1)

# 6️⃣ Keep only the country and the average
mobile_avg = mobile[['Country Name', 'Avg_Mobile_Subscriptions']]

# 7️⃣ Rename for consistency with other datasets
mobile_avg.rename(columns={'Country Name': 'Entity'}, inplace=True)

# 8️⃣ Save the clean dataset
mobile_avg.to_excel("Avg_Mobile_Subscriptions.xlsx", index=False)

print("✅ Done! 'Avg_Mobile_Subscriptions.xlsx' created with averages from 2000–2023.")
```

```python
#Cleaning the Human Development Index (HDI) dataset and calculating average HDI per country from 2000 to 2023, then saving to Excel.
import pandas as pd

# 1️⃣ Load the dataset
hdi = pd.read_csv("human-development-index.csv")

# 2️⃣ Inspect the first few rows
print(hdi.head())

# 3️⃣ Standardize column names
hdi.columns = [col.strip().lower().replace(" ", "_") for col in hdi.columns]

# 4️⃣ If your file has columns like "Country", "Year", "HDI"
#    we’ll group by country and average the HDI from 2000–2023
if {"country", "year", "hdi"}.issubset(hdi.columns):
    # Keep only years 2000–2023
    hdi_filtered = hdi[(hdi["year"] >= 2000) & (hdi["year"] <= 2023)]
    
    # Compute average per country
    avg_hdi = hdi_filtered.groupby("country", as_index=False)["hdi"].mean()
    avg_hdi = avg_hdi.rename(columns={"hdi": "Average_HDI"})
    
    # Save and preview
    avg_hdi.to_excel("Avg_HDI.xlsx", index=False)
    print(avg_hdi.head())

else:
    print("⚠️ Column names don't match expected structure. Please print hdi.head() to inspect.")
```

```python

#Cleaning the Human Development Index (HDI) dataset and calculating average HDI per country from 2000 to 2023, then saving to Excel.
import pandas as pd

# 1️⃣ Load the dataset
hdi = pd.read_csv("human-development-index.csv")

# 2️⃣ Filter years between 2000 and 2023
hdi_filtered = hdi[(hdi["Year"] >= 2000) & (hdi["Year"] <= 2023)]

# 3️⃣ Group by country (Entity) and calculate mean HDI
hdi_avg = (
    hdi_filtered.groupby("Entity")["Human Development Index"]
    .mean()
    .reset_index()
    .rename(columns={"Entity": "Country", "Human Development Index": "Average_HDI"})
)

# 4️⃣ Save the results to Excel
hdi_avg.to_excel("Avg_HDI.xlsx", index=False)

print("✅ Done! File saved as 'Avg_HDI.xlsx'")
print(hdi_avg.head())
```

```python
import pandas as pd

# 1Load all datasets
gdp = pd.read_excel("Avg_GDP.xlsx")
internet = pd.read_excel("Avg_Internet_Usage.xlsx")
mobile = pd.read_excel("Avg_Mobile_Subscriptions.xlsx")
hdi = pd.read_excel("Avg_HDI.xlsx")

# 2 Rename "Entity" to "Country" where needed
gdp.rename(columns={"Entity": "Country"}, inplace=True)
internet.rename(columns={"Entity": "Country"}, inplace=True)
mobile.rename(columns={"Entity": "Country"}, inplace=True)

# 3 Merge everything on "Country"
merged = (
    gdp.merge(internet, on="Country", how="outer")
       .merge(mobile, on="Country", how="outer")
       .merge(hdi, on="Country", how="outer")
)

# 4 Optional cleanup and sorting
merged = merged.sort_values(by="Country").reset_index(drop=True)

# 5 Save to Excel
merged.to_excel("Global_Data_Merged.xlsx", index=False)

print(" Successfully merged! File saved as 'Global_Data_Merged.xlsx'")
print(merged.head())
```

```python
import pandas as pd
import plotly.express as px
import plotly.io as pio

# Force rendering in browser (avoids nbformat issues)
pio.renderers.default = "browser"

# Load cleaned dataset
df = pd.read_excel("Global_Data_Merged_Cleaned.xlsx")

# Check column exists
if 'Avg_Internet_Usage' not in df.columns:
    raise ValueError("⚠️ Column 'Avg_Internet_Usage' not found!")

# Top 10 and bottom 10 countries
top10 = df.sort_values(by='Avg_Internet_Usage', ascending=False).head(10)
bottom10 = df.sort_values(by='Avg_Internet_Usage', ascending=True).head(10)

# Combine for plotting
combined = pd.concat([top10, bottom10])
combined['Group'] = ['Top 10']*10 + ['Bottom 10']*10

# Create horizontal bar chart
fig = px.bar(
    combined,
    x='Avg_Internet_Usage',
    y='Country',
    color='Group',
    orientation='h',
    text='Avg_Internet_Usage',
    title="🌐 Top 10 and Bottom 10 Countries by Average Internet Usage (2000–2023)",
    color_discrete_map={'Top 10':'green', 'Bottom 10':'red'}
)

# Layout tweaks
fig.update_layout(
    yaxis=dict(autorange="reversed"),
    title_x=0.5,
    title_font_size=16
)

# Show chart in browser
fig.show()

# Optional: save to HTML
fig.write_html("Top_Bottom10_Internet_Chart.html")
print("✅ Chart created and saved as HTML! Open in browser to view it.")
```

```python
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
import plotly.io as pio

# 1️⃣ Force Plotly to render in browser
pio.renderers.default = "browser"

# 2️⃣ Load the cleaned dataset
df = pd.read_excel("Global_Data_Merged_Cleaned.xlsx")

# 3️⃣ Check columns
required_cols = ['Avg_Internet_Usage', 'Average_HDI', 'Country']
for col in required_cols:
    if col not in df.columns:
        raise ValueError(f"⚠️ Column '{col}' not found!")

# 4️⃣ Scatter plot of Internet vs HDI
fig = px.scatter(
    df,
    x='Avg_Internet_Usage',
    y='Average_HDI',
    hover_name='Country',
    size='Avg_Internet_Usage',
    color='Average_HDI',
    color_continuous_scale='Viridis',
    title='🌐 Internet Penetration vs Human Development Index (2000–2023)',
    labels={'Avg_Internet_Usage':'Average Internet Usage (%)', 'Average_HDI':'Average HDI'}
)

# 5️⃣ Add a linear regression line manually using NumPy
x = df['Avg_Internet_Usage'].to_numpy()
y = df['Average_HDI'].to_numpy()

# Linear regression: y = m*x + b
m, b = np.polyfit(x, y, 1)
y_pred = m*x + b

# Add trendline to the plot
fig.add_trace(go.Scatter(x=x, y=y_pred, mode='lines', name='Trendline', line=dict(color='red')))

# 6️⃣ Show the plot
fig.show()

# 7️⃣ Save as HTML
fig.write_html("Internet_vs_HDI.html")
print("✅ Scatter plot with trendline created and saved as HTML. Open in browser to view it.")
```
